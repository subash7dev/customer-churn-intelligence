```python
# ============================================================
# CUSTOMER CHURN INTELLIGENCE
# 02 - EXPLORATORY DATA ANALYSIS
# ============================================================

# %% [markdown]
# # Customer Churn Intelligence
# ## 02. Exploratory Data Analysis
#
# Objective:
# - Understand customer behavior
# - Identify churn patterns
# - Analyze customer segments
# - Identify revenue risk
# - Generate business insights for ML modeling

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")

# %% [markdown]
# ## 1. Load Clean Dataset

# %%
file_path = "../data/processed/telco_customer_churn_clean.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

display(df.head())

# %% [markdown]
# ## 2. Basic Dataset Information

# %%
df.info()

# %% [markdown]
# ## 3. Churn Overview

# %%
churn_counts = df["Churn"].value_counts()

print("Churn counts:")
display(churn_counts)

# %%
churn_percentage = (
    df["Churn"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Churn percentage:")
display(churn_percentage)

# %% [markdown]
# ## 4. Churn Visualization

# %%
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.xticks(
    [0, 1],
    ["Stayed", "Churned"]
)

plt.show()

# %% [markdown]
# ## 5. Gender vs Churn

# %%
gender_churn = pd.crosstab(
    df["gender"],
    df["Churn"],
    normalize="index"
) * 100

display(gender_churn.round(2))

# %%
gender_churn.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Churn Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Percentage")

plt.xticks(rotation=0)
plt.legend(["Stayed", "Churned"])

plt.show()

# %% [markdown]
# ## 6. Senior Citizen vs Churn

# %%
senior_churn = pd.crosstab(
    df["SeniorCitizen"],
    df["Churn"],
    normalize="index"
) * 100

display(senior_churn.round(2))

# %%
sns.barplot(
    data=df,
    x="SeniorCitizen",
    y="Churn"
)

plt.title("Churn Rate by Senior Citizen Status")
plt.xlabel("Senior Citizen")
plt.ylabel("Churn Rate")

plt.xticks(
    [0, 1],
    ["Non-Senior", "Senior"]
)

plt.show()

# %% [markdown]
# ## 7. Partner Status vs Churn

# %%
partner_churn = pd.crosstab(
    df["Partner"],
    df["Churn"],
    normalize="index"
) * 100

display(partner_churn.round(2))

# %%
sns.barplot(
    data=df,
    x="Partner",
    y="Churn"
)

plt.title("Churn Rate by Partner Status")
plt.xlabel("Partner")
plt.ylabel("Churn Rate")

plt.xticks(
    [0, 1],
    ["No Partner", "Has Partner"]
)

plt.show()

# %% [markdown]
# ## 8. Dependents vs Churn

# %%
dependents_churn = pd.crosstab(
    df["Dependents"],
    df["Churn"],
    normalize="index"
) * 100

display(dependents_churn.round(2))

# %%
sns.barplot(
    data=df,
    x="Dependents",
    y="Churn"
)

plt.title("Churn Rate by Dependents")
plt.xlabel("Dependents")
plt.ylabel("Churn Rate")

plt.xticks(
    [0, 1],
    ["No Dependents", "Has Dependents"]
)

plt.show()

# %% [markdown]
# ## 9. Tenure Distribution

# %%
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="tenure",
    bins=30,
    kde=True
)

plt.title("Customer Tenure Distribution")
plt.xlabel("Tenure (Months)")
plt.ylabel("Number of Customers")

plt.show()

# %% [markdown]
# ## 10. Tenure vs Churn

# %%
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)

plt.title("Tenure Distribution by Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")

plt.xticks(
    [0, 1],
    ["Stayed", "Churned"]
)

plt.show()

# %% [markdown]
# ## 11. Create Tenure Groups

# %%
df["TenureGroup"] = pd.cut(
    df["tenure"],
    bins=[-1, 6, 12, 24, 48, 72],
    labels=[
        "0-6 Months",
        "7-12 Months",
        "13-24 Months",
        "25-48 Months",
        "49-72 Months"
    ]
)

# %%
tenure_churn = pd.crosstab(
    df["TenureGroup"],
    df["Churn"],
    normalize="index"
) * 100

display(tenure_churn.round(2))

# %%
sns.barplot(
    data=df,
    x="TenureGroup",
    y="Churn"
)

plt.title("Churn Rate by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Churn Rate")

plt.xticks(rotation=30)

plt.show()

# %% [markdown]
# ## 12. Contract Type vs Churn

# %%
contract_churn = pd.crosstab(
    df["Contract"],
    df["Churn"],
    normalize="index"
) * 100

display(contract_churn.round(2))

# %%
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="Contract",
    y="Churn"
)

plt.title("Churn Rate by Contract Type")
plt.xlabel("Contract")
plt.ylabel("Churn Rate")

plt.xticks(rotation=15)

plt.show()

# %% [markdown]
# ## 13. Internet Service vs Churn

# %%
internet_churn = pd.crosstab(
    df["InternetService"],
    df["Churn"],
    normalize="index"
) * 100

display(internet_churn.round(2))

# %%
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="InternetService",
    y="Churn"
)

plt.title("Churn Rate by Internet Service")
plt.xlabel("Internet Service")
plt.ylabel("Churn Rate")

plt.show()

# %% [markdown]
# ## 14. Payment Method vs Churn

# %%
payment_churn = pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100

display(payment_churn.round(2))

# %%
plt.figure(figsize=(11, 5))

sns.barplot(
    data=df,
    x="PaymentMethod",
    y="Churn"
)

plt.title("Churn Rate by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Churn Rate")

plt.xticks(rotation=30, ha="right")

plt.show()

# %% [markdown]
# ## 15. Monthly Charges vs Churn

# %%
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges by Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")

plt.xticks(
    [0, 1],
    ["Stayed", "Churned"]
)

plt.show()

# %% [markdown]
# ## 16. Total Charges vs Churn

# %%
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="TotalCharges"
)

plt.title("Total Charges by Churn")
plt.xlabel("Churn")
plt.ylabel("Total Charges")

plt.xticks(
    [0, 1],
    ["Stayed", "Churned"]
)

plt.show()

# %% [markdown]
# ## 17. Additional Services vs Churn

# %%
service_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

service_results = []

for column in service_columns:

    temp = pd.crosstab(
        df[column],
        df["Churn"],
        normalize="index"
    ) * 100

    for category in temp.index:

        service_results.append({
            "Service": column,
            "Category": category,
            "ChurnRate": temp.loc[category, 1]
        })

service_churn = pd.DataFrame(service_results)

display(
    service_churn.sort_values(
        "ChurnRate",
        ascending=False
    )
)

# %% [markdown]
# ## 18. Monthly Revenue Analysis

# %%
total_monthly_revenue = df["MonthlyCharges"].sum()

churned_monthly_revenue = df.loc[
    df["Churn"] == 1,
    "MonthlyCharges"
].sum()

retained_monthly_revenue = df.loc[
    df["Churn"] == 0,
    "MonthlyCharges"
].sum()

print(
    f"Total monthly revenue: ${total_monthly_revenue:,.2f}"
)

print(
    f"Monthly revenue from churned customers: "
    f"${churned_monthly_revenue:,.2f}"
)

print(
    f"Monthly revenue from retained customers: "
    f"${retained_monthly_revenue:,.2f}"
)

# %% [markdown]
# ## 19. Revenue at Risk

# %%
revenue_at_risk_percentage = (
    churned_monthly_revenue /
    total_monthly_revenue
) * 100

print(
    f"Monthly revenue associated with churned customers: "
    f"{revenue_at_risk_percentage:.2f}%"
)

# %% [markdown]
# ## 20. Average Customer Metrics by Churn

# %%
customer_metrics = df.groupby("Churn").agg(
    Customers=("customerID", "count"),
    AvgTenure=("tenure", "mean"),
    AvgMonthlyCharges=("MonthlyCharges", "mean"),
    AvgTotalCharges=("TotalCharges", "mean")
).round(2)

display(customer_metrics)

# %% [markdown]
# ## 21. Contract + Internet Service Analysis

# %%
contract_internet = (
    df.groupby(
        ["Contract", "InternetService"]
    )
    .agg(
        Customers=("customerID", "count"),
        ChurnRate=("Churn", "mean"),
        AvgMonthlyCharges=("MonthlyCharges", "mean")
    )
)

contract_internet["ChurnRate"] *= 100

contract_internet = contract_internet.round(2)

display(
    contract_internet.sort_values(
        "ChurnRate",
        ascending=False
    )
)

# %% [markdown]
# ## 22. High-Risk Customer Segment

# %%
high_risk = df[
    (df["Contract"] == "Month-to-month") &
    (df["tenure"] <= 12) &
    (df["MonthlyCharges"] >= df["MonthlyCharges"].median())
]

print("High-risk segment size:", len(high_risk))

print(
    "High-risk segment percentage:",
    round(len(high_risk) / len(df) * 100, 2),
    "%"
)

print(
    "High-risk segment churn rate:",
    round(high_risk["Churn"].mean() * 100, 2),
    "%"
)

# %% [markdown]
# ## 23. Correlation Analysis
#
# Convert selected categorical variables into numeric form
# for correlation analysis.

# %%
correlation_df = df.copy()

correlation_df["gender"] = correlation_df["gender"].map({
    "Male": 1,
    "Female": 0
})

# %%
numeric_corr = correlation_df.select_dtypes(
    include=["int64", "float64"]
)

correlation_matrix = numeric_corr.corr()

display(
    correlation_matrix["Churn"]
    .sort_values(ascending=False)
)

# %% [markdown]
# ## 24. Correlation Heatmap

# %%
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.show()

# %% [markdown]
# ## 25. Top Churn Indicators
#
# This is exploratory analysis, not causal inference.

# %%
churn_correlation = (
    correlation_matrix["Churn"]
    .drop("Churn")
    .sort_values(
        key=abs,
        ascending=False
    )
)

display(churn_correlation)

# %% [markdown]
# ## 26. Save EDA Summary Tables

# %%
contract_churn.to_csv(
    "../data/processed/contract_churn_analysis.csv"
)

internet_churn.to_csv(
    "../data/processed/internet_churn_analysis.csv"
)

payment_churn.to_csv(
    "../data/processed/payment_churn_analysis.csv"
)

service_churn.to_csv(
    "../data/processed/service_churn_analysis.csv",
    index=False
)

print("EDA summary tables saved.")

# %% [markdown]
# # EDA Completed
#
# Key areas analyzed:
#
# - Overall churn
# - Customer demographics
# - Tenure
# - Contract type
# - Internet service
# - Payment method
# - Additional services
# - Monthly charges
# - Total charges
# - Revenue at risk
# - Customer segments
# - Correlation indicators
#
# Next:
# Feature Engineering → Machine Learning → Churn Prediction
```
